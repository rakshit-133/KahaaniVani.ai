```python
import os, shutil, torch

print(f"GPUs: {torch.cuda.device_count()}")
os.system("pip install transformers datasets scikit-learn accelerate tqdm -q")

# Copy the file from your dataset to the working directory
shutil.copy("/kaggle/input/datasets/rakshit133/kahaanivani-training-data/finetune.py",
            "/kaggle/working/finetune.py")

# Apply the aggressive tuning patches directly in Kaggle!
with open("/kaggle/working/finetune.py", "r") as f:
    code = f.read()
code = code.replace('"epochs":         8,', '"epochs":         4,')
code = code.replace('"learning_rate":  1e-5,', '"learning_rate":  5e-5,')
code = code.replace('"warmup_ratio":   0.15,', '"warmup_ratio":   0.1,')
code = code.replace('"use_weighted_loss": True,', '"use_weighted_loss": False,')
with open("/kaggle/working/finetune.py", "w") as f:
    f.write(code)

os.makedirs("/kaggle/working/data", exist_ok=True)
shutil.copy("/kaggle/input/datasets/rakshit133/kahaanivani-training-data/combined_train.csv", "/kaggle/working/data/combined_train.csv")
shutil.copy("/kaggle/input/datasets/rakshit133/kahaanivani-training-data/combined_val.csv",   "/kaggle/working/data/combined_val.csv")
os.chdir("/kaggle/working")

# Run the aggressive training
os.system("python finetune.py")

# Zip and download
shutil.make_archive("/kaggle/working/emotion_model_aggressive", "zip", "/kaggle/working/output/emotion_model")
print("\n✅ DONE! Download emotion_model_aggressive.zip from Output panel on the right ->")
```
